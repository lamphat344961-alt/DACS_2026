```python
import pandas as pd
files = "C:\\Users\\Admin\\Desktop\\dataset_demo.xls"
# 1. Danh sách các sheet cần đọc
sheet_names = ["VIDEOS", "TAGS", "REPLIES", "CREATOR_TAGS", "COMMENTS"]

# 2. Hàm lọc cột có hơn 80% Null (hàm bạn đã yêu cầu trước đó)
def filter_null_columns(df, threshold=0.8):
    null_pct = df.isnull().mean()
    cols_to_drop = null_pct[null_pct > threshold].index.tolist()
    return df.drop(columns=cols_to_drop)

# 3. Tạo Dictionary để lưu các DataFrame mới sau khi lọc
cleaned_dfs = {}

# 4. Vòng lặp qua từng sheet
for sheet in sheet_names:
    print(f"--- Đang xử lý sheet: {sheet} ---")
    
    # Đọc dữ liệu từ file Excel
    # Lưu ý: Thêm r trước đường dẫn để tránh lỗi unicodeescape
    df = pd.read_excel(r"C:\Users\Admin\Desktop\dataset_demo.xls", sheet_name=sheet)
    
    # Áp dụng hàm lọc
    df_new = filter_null_columns(df, threshold=0.8)
    
    # Lưu vào dictionary với key là tên sheet (ví dụ: cleaned_dfs['VIDEOS'])
    cleaned_dfs[sheet] = df_new
    
    print(f"Kích thước sau khi lọc: {df_new.shape}\n")

df_creators = pd.read_excel(files, sheet_name="CREATORS")
```

```python
# 1. Gán các biến riêng lẻ từ dictionary (nếu bạn cần dùng code sau này)
df_videos_new = cleaned_dfs["VIDEOS"]
df_tags_new = cleaned_dfs["TAGS"]
df_replies_new = cleaned_dfs["REPLIES"]
df_creators
df_creator_tags_new = cleaned_dfs["CREATOR_TAGS"]
df_comments_new = cleaned_dfs["COMMENTS"]

# 2. Tạo đường dẫn lưu file (sử dụng r'' để tránh lỗi unicode)
output_path = r"C:\\Users\\Admin\\Desktop\\TANPHAT\\Dự án_canhan\\DACS\\dataset.xlsx"

# 3. Sử dụng ExcelWriter để ghi nhiều sheet vào 1 file
with pd.ExcelWriter(output_path) as writer:
    df_videos_new.to_excel(writer, sheet_name="VIDEOS", index=False)
    df_tags_new.to_excel(writer, sheet_name="TAGS", index=False)
    df_replies_new.to_excel(writer, sheet_name="REPLIES", index=False)
    df_creators.to_excel(writer, sheet_name="CREATORS", index=False)
    df_creator_tags_new.to_excel(writer, sheet_name="CREATOR_TAGS", index=False)
    df_comments_new.to_excel(writer, sheet_name="COMMENTS", index=False)

print(f"Đã lưu thành công file tại: {output_path}")
```
