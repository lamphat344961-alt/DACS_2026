## 0. Imports

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from itertools import product
from mpl_toolkits.mplot3d import Axes3D

from sklearn.preprocessing import RobustScaler
from sklearn.decomposition import KernelPCA, PCA
from sklearn.cluster import KMeans, AgglomerativeClustering, DBSCAN, SpectralClustering
from sklearn.metrics import silhouette_score, davies_bouldin_score, calinski_harabasz_score
import scipy.cluster.hierarchy as sch

sns.set_theme(style="whitegrid")
# ==============================
# Cấu hình hình ảnh cho báo cáo
# ==============================


FIG_SIZE = (10, 6)   # kích thước hình, đơn vị inch
DPI = 300            # độ nét khi xuất hình

plt.rcParams["figure.figsize"] = FIG_SIZE
plt.rcParams["figure.dpi"] = DPI
plt.rcParams["savefig.dpi"] = DPI
```

## 1. Load dữ liệu · PCA


```python
df_raw = pd.read_csv('pipeline_P6_scaled.csv')

handle_col = df_raw['handleName']

df = df_raw.drop(columns=['handleName'])

pca = PCA(n_components=0.95)
X_model = pca.fit_transform(df)
print(f"Số chiều sau PCA: {X_model.shape[1]}")
```

## 2. K-Means

### 2a. Elbow Method – WCSS (k = 1 → 10)

```python
wcss = []
K_range = range(1, 11)
for k in K_range:
    km = KMeans(n_clusters=k, random_state=42, n_init=10)
    km.fit(X_model)
    wcss.append(km.inertia_)

plt.figure(figsize=(8, 4))
plt.plot(list(K_range), wcss, marker='o', linewidth=2)
plt.xlabel('Số cụm (K)')
plt.ylabel('WCSS')
plt.title('Elbow Method')
plt.xticks(list(K_range))
plt.tight_layout()
plt.show()
```

### 2b. Silhouette Score (k = 2 → 10) – chọn k có score cao nhất

```python
K_sil = range(2, 11)
sil_km = []
for k in K_sil:
    labels = KMeans(n_clusters=k, random_state=42, n_init=10).fit_predict(X_model)
    sil_km.append(silhouette_score(X_model, labels))

best_k_km = list(K_sil)[int(np.argmax(sil_km))]

plt.figure(figsize=(8, 4))
plt.plot(list(K_sil), sil_km, marker='o', linewidth=2)
plt.axvline(best_k_km, color='red', linestyle='--', label=f'Best K = {best_k_km}')
plt.xlabel('Số cụm (K)')
plt.ylabel('Silhouette Score')
plt.title('Silhouette Score – K-Means')
plt.legend()
plt.tight_layout()
plt.show()
print(f"K tối ưu (Silhouette): {best_k_km}")
```

### 2c. Fit K-Means với k tối ưu

```python
K_KMEANS     = best_k_km   # ← ghi đè thủ công nếu muốn
kmeans       = KMeans(n_clusters=K_KMEANS, random_state=42, n_init=10)
labels_kmeans = kmeans.fit_predict(X_model)
print(f"K-Means fitted với K = {K_KMEANS}")


# ── K-Means với K = 8 ─────────────────────────────────────────────────────
kmeans_8      = KMeans(n_clusters=8, random_state=42, n_init=10)
print("K-Means fitted với K = 8")
labels_kmeans_8 = kmeans_8.fit_predict(X_model)
```

## 3. Hierarchical Clustering (Ward)

### 3a. Dendrogram

```python
plt.figure(figsize=(13, 6))
sch.dendrogram(
    sch.linkage(X_model, method='ward'),
    truncate_mode='lastp', p=30,
    leaf_rotation=45, leaf_font_size=9
)
plt.title('Dendrogram – Hierarchical Clustering (Ward)')
plt.xlabel('Điểm dữ liệu')
plt.ylabel('Khoảng cách Ward')
plt.tight_layout()
plt.show()
```

### 3b. Silhouette Score theo khoảng k tùy chọn

```python
K_HC_RANGE = range(2, 9)   # ← điều chỉnh khoảng k tại đây

sil_hc = []
for k in K_HC_RANGE:
    labels = AgglomerativeClustering(n_clusters=k, linkage='ward').fit_predict(X_model)
    sil_hc.append(silhouette_score(X_model, labels))

best_k_hc = list(K_HC_RANGE)[int(np.argmax(sil_hc))]

plt.figure(figsize=(8, 4))
plt.plot(list(K_HC_RANGE), sil_hc, marker='o', linewidth=2)
plt.axvline(best_k_hc, color='red', linestyle='--', label=f'Best K = {best_k_hc}')
plt.xlabel('Số cụm (K)')
plt.ylabel('Silhouette Score')
plt.title('Silhouette Score – Hierarchical Clustering')
plt.legend()
plt.tight_layout()
plt.show()
print(f"K tối ưu Hierarchical: {best_k_hc}")
```

### 3c. Fit Hierarchical với k tối ưu

```python
K_HC    = 4       # ← ghi đè thủ công nếu muốn
hc_model = AgglomerativeClustering(n_clusters=K_HC, linkage='ward')
labels_hc = hc_model.fit_predict(X_model)
print(f"Hierarchical fitted với K = {K_HC}")
```

## 4. DBSCAN – Grid Search tìm tham số tối ưu

### 4a. Grid Search (eps × min_samples) – tối đa hoá Silhouette Score

```python
# ── Grid rộng để đảm bảo tìm được cấu hình hợp lệ ───────────────────────
eps_grid         = [ 0.5, 0.8, 1.0, 1.5, 2.0, 3.0, 5.0, 8.0, 10.0]
min_samples_grid = [3, 5, 7, 9, 11, 15]
NOISE_RATIO_MAX  = 0.6   # ← cho phép tối đa 60% noise, tăng nếu cần

best_db_score  = -1
best_db_params = {}
records_db     = []

for eps, ms in product(eps_grid, min_samples_grid):
    labels  = DBSCAN(eps=eps, min_samples=ms).fit_predict(X_model)
    n_clust = len(set(labels)) - (1 if -1 in labels else 0)
    n_noise = int(np.sum(labels == -1))
    noise_ratio = n_noise / len(labels)

    if n_clust >= 2 and noise_ratio < NOISE_RATIO_MAX:
        score = silhouette_score(X_model, labels)
        records_db.append({
            'eps': eps, 'min_samples': ms,
            'n_clusters': n_clust, 'n_noise': n_noise,
            'noise_pct': round(noise_ratio * 100, 1),
            'silhouette': round(score, 4)
        })
        if score > best_db_score:
            best_db_score  = score
            best_db_params = {'eps': eps, 'min_samples': ms}

if not records_db:
    print("⚠️  Không tìm được cấu hình DBSCAN hợp lệ!")
    print("   → Thử tăng NOISE_RATIO_MAX hoặc mở rộng eps_grid / min_samples_grid")
    best_db_params = {'eps': eps_grid[len(eps_grid)//2],
                      'min_samples': min_samples_grid[0]}
else:
    df_db_grid = pd.DataFrame(records_db).sort_values('silhouette', ascending=False)
    print(df_db_grid.head(10).to_string(index=False))
    print(f"\nTham số tối ưu: eps={best_db_params['eps']}, "
          f"min_samples={best_db_params['min_samples']}, "
          f"Silhouette={best_db_score:.4f}")
```

```python
import os

# Tạo thư mục lưu hình
FIG_DIR = "report_figures"
os.makedirs(FIG_DIR, exist_ok=True)

# Nếu df_db_grid chưa tồn tại nhưng records_db có dữ liệu
if "df_db_grid" not in globals():
    if len(records_db) > 0:
        df_db_grid = pd.DataFrame(records_db)
    else:
        df_db_grid = pd.DataFrame()

if df_db_grid.empty:
    print("⚠️ Không có kết quả DBSCAN hợp lệ để vẽ heatmap.")
else:
    # Đảm bảo đúng kiểu dữ liệu
    df_db_grid["eps"] = df_db_grid["eps"].astype(float)
    df_db_grid["min_samples"] = df_db_grid["min_samples"].astype(int)

    def plot_dbscan_heatmap(
        data,
        value_col,
        title,
        filename,
        fmt=".2f",
        figsize=(10, 6),
        dpi=300
    ):
        """
        Vẽ heatmap cho kết quả grid search DBSCAN.
        Hàng: min_samples
        Cột: eps
        Giá trị: value_col
        """

        heatmap_data = data.pivot(
            index="min_samples",
            columns="eps",
            values=value_col
        )

        fig, ax = plt.subplots(figsize=figsize, dpi=dpi)

        im = ax.imshow(
            heatmap_data.values,
            aspect="auto"
        )

        # Gắn nhãn trục
        ax.set_xticks(np.arange(len(heatmap_data.columns)))
        ax.set_yticks(np.arange(len(heatmap_data.index)))

        ax.set_xticklabels(heatmap_data.columns)
        ax.set_yticklabels(heatmap_data.index)

        ax.set_xlabel("eps")
        ax.set_ylabel("min_samples")
        ax.set_title(title)

        # Ghi giá trị lên từng ô
        for i in range(len(heatmap_data.index)):
            for j in range(len(heatmap_data.columns)):
                val = heatmap_data.iloc[i, j]
                if pd.notna(val):
                    ax.text(
                        j, i,
                        format(val, fmt),
                        ha="center",
                        va="center",
                        fontsize=9
                    )

        # Thanh màu
        cbar = fig.colorbar(im, ax=ax)
        cbar.set_label(value_col)

        plt.tight_layout()

        # Lưu PNG cho Word
        plt.savefig(
            os.path.join(FIG_DIR, filename + ".png"),
            dpi=dpi,
            bbox_inches="tight",
            facecolor="white"
        )

        # Lưu PDF vector cho báo cáo/PDF
        plt.savefig(
            os.path.join(FIG_DIR, filename + ".pdf"),
            bbox_inches="tight",
            facecolor="white"
        )

        plt.show()

    # 1. Heatmap Silhouette
    plot_dbscan_heatmap(
        data=df_db_grid,
        value_col="silhouette",
        title="DBSCAN Grid Search - Silhouette Score",
        filename="heatmap_dbscan_silhouette",
        fmt=".4f",
        figsize=(11, 6)
    )

    # 2. Heatmap số cụm
    plot_dbscan_heatmap(
        data=df_db_grid,
        value_col="n_clusters",
        title="DBSCAN Grid Search - Number of Clusters",
        filename="heatmap_dbscan_n_clusters",
        fmt=".0f",
        figsize=(11, 6)
    )

    # 3. Heatmap tỷ lệ noise
    plot_dbscan_heatmap(
        data=df_db_grid,
        value_col="noise_pct",
        title="DBSCAN Grid Search - Noise Percentage (%)",
        filename="heatmap_dbscan_noise_pct",
        fmt=".1f",
        figsize=(11, 6)
    )
```

## 5. Spectral Clustering – Ma trận Silhouette Score

### 5a. Grid Search: k × sigma × k_nn  
`k_clusters` = 2–6 · `sigma` = [0.3, 0.5, 0.7, 1.0] · `k_nn` = [5, 10, 15]

```python
# Tìm k_nn nhỏ nhất mà fully connected
import warnings
for k_nn in [10, 15, 20, 30, 50]:
    sc = SpectralClustering(n_clusters=3, affinity='nearest_neighbors',
                            n_neighbors=k_nn, random_state=42)
    with warnings.catch_warnings(record=True) as w:
        warnings.simplefilter("always")
        sc.fit_predict(X_model)
        connected = len(w) == 0
    print(f"k_nn={k_nn:3d} → {'✓ connected' if connected else '✗ not connected'}")
```

```python
#  dùng affinity='nearest_neighbors' (sparse) + subsample grid search.

import time

k_clusters_grid = range(2, 7)       # k = 2, 3, 4, 5, 6
k_nn_grid       = [30,35,40,42,45,47,50,55,60]   # số láng giềng cho nearest_neighbors affinity

# ── Subsample để grid search nhanh ───────────────────────────────────────
SUBSAMPLE_N = 5089    # ← tăng nếu dataset nhỏ hơn 2000; giảm nếu vẫn chậm
rng = np.random.default_rng(42)
idx_sub = rng.choice(len(X_model), size=min(SUBSAMPLE_N, len(X_model)), replace=False)
X_sub   = X_model[idx_sub]
print(f"Grid search trên {len(X_sub)} mẫu (/{len(X_model)} tổng)")

records_sc = []
t0 = time.time()

for k_nn in k_nn_grid:
    row = {'k_nn': k_nn}
    for k in k_clusters_grid:
        try:
            sc     = SpectralClustering(
                         n_clusters  = k,
                         affinity    = 'nearest_neighbors',  # sparse – nhanh hơn rbf rất nhiều
                         n_neighbors = k_nn,
                         n_jobs      = -1,                   # dùng toàn bộ CPU
                         random_state = 42,
                         n_init      = 10
                     )
            labels = sc.fit_predict(X_sub)
            score  = silhouette_score(X_sub, labels) if len(set(labels)) > 1 else np.nan
        except Exception as e:
            score = np.nan
        row[f'k={k}'] = round(float(score), 4) if (score is not None and score == score) else np.nan
    records_sc.append(row)
    elapsed = time.time() - t0
    print(f"  k_nn={k_nn:2d} xong — {elapsed:.1f}s")

df_spec_grid = pd.DataFrame(records_sc)
print("\n── Ma trận Silhouette Score ─────────────────────────────────────────")
print(df_spec_grid.to_string(index=False))

# ── Tìm tham số tốt nhất ─────────────────────────────────────────────────
k_cols   = [f'k={k}' for k in k_clusters_grid]
stacked  = df_spec_grid[k_cols].stack().dropna()
best_loc = stacked.idxmax()                     # (row_idx, 'k=X')
best_sc_row = df_spec_grid.loc[best_loc[0]]
best_sc_k   = int(best_loc[1].split('=')[1])
best_sc_knn = int(best_sc_row['k_nn'])
print(f"\nTốt nhất: k_nn={best_sc_knn}, k={best_sc_k}, "
      f"Silhouette={stacked[best_loc]:.4f}")
print(f"Tổng thời gian grid search: {time.time() - t0:.1f}s")
```

```python
# ============================================================
# HEATMAP SPECTRAL CLUSTERING GRID SEARCH
# ============================================================

import numpy as np
import matplotlib.pyplot as plt

# Các cột điểm silhouette: k=2, k=3, ...
k_cols = [f'k={k}' for k in k_clusters_grid]

# Ma trận heatmap
heatmap_data = df_spec_grid.set_index("k_nn")[k_cols]

fig, ax = plt.subplots()

im = ax.imshow(
    heatmap_data.values,
    aspect="auto"
)

# Nhãn trục
ax.set_xticks(np.arange(len(heatmap_data.columns)))
ax.set_yticks(np.arange(len(heatmap_data.index)))

ax.set_xticklabels(heatmap_data.columns)
ax.set_yticklabels(heatmap_data.index)

ax.set_xlabel("Số cụm k")
ax.set_ylabel("Số láng giềng k_nn")
ax.set_title("Heatmap Silhouette Score - Spectral Clustering")

# Ghi giá trị silhouette lên từng ô
for i in range(len(heatmap_data.index)):
    for j in range(len(heatmap_data.columns)):
        value = heatmap_data.iloc[i, j]
        if not np.isnan(value):
            ax.text(
                j, i,
                f"{value:.4f}",
                ha="center",
                va="center",
                fontsize=9
            )

# Thanh màu
cbar = fig.colorbar(im, ax=ax)
cbar.set_label("Silhouette Score")

plt.tight_layout()

# Lưu hình
plt.savefig(
    "heatmap_spectral_silhouette.png",
    dpi=DPI,
    bbox_inches="tight"
)

plt.show()
```

```python
K_SPECTRAL   = 2
KNN_SPECTRAL = 30

spectral = SpectralClustering(
    n_clusters   = K_SPECTRAL,
    affinity     = 'nearest_neighbors',
    n_neighbors  = KNN_SPECTRAL,
    random_state = 42,
    n_init       = 10,
    n_jobs       = -1
)
labels_spec = spectral.fit_predict(X_model)
```

## 6. Tổng kết – Silhouette · DBI · CHI

```python
def evaluate(labels, name):
    mask   = labels != -1          # loại bỏ noise của DBSCAN
    X_eval = X_model[mask]
    l_eval = labels[mask]
    n_clust = len(set(l_eval))
    if n_clust < 2:
        return {'Algorithm': name, 'N_Clusters': n_clust,
                'Silhouette': np.nan, 'DBI': np.nan, 'CHI': np.nan}
    return {
        'Algorithm'  : name,
        'N_Clusters' : n_clust,
        'Silhouette' : round(silhouette_score(X_eval, l_eval), 4),
        'DBI'        : round(davies_bouldin_score(X_eval, l_eval), 4),
        'CHI'        : round(calinski_harabasz_score(X_eval, l_eval), 2),
    }

df_summary = pd.DataFrame([
    evaluate(labels_kmeans, 'K-Means'),
    evaluate(labels_hc,     'Hierarchical'),
    evaluate(labels_db,     'DBSCAN'),
    evaluate(labels_spec,   'Spectral'),

])
print(df_summary.to_string(index=False))
```

### 10a. Heatmap DBSCAN Grid Search

Heatmap này dùng lại `records_db`/`df_db_grid` đã tạo ở phần DBSCAN. Không fit lại model và không đổi tham số DBSCAN đã chọn.

### 10c. Xuất lại các hình so sánh cụm 2D/3D chất lượng cao

Cell này vẽ lại các hình PCA 2D và PCA 3D đã có, sau đó lưu vào thư mục `report_figures`. Không thay đổi nhãn cụm.


```python
# ── Xuất hình so sánh cụm 2D PCA chất lượng cao ─────────────────────────
# Dùng lại all_models và X_2d đã tạo ở phần trực quan hóa.

if "X_2d" not in globals():
    pca2d = PCA(n_components=2, random_state=42)
    X_2d = pca2d.fit_transform(X_model)

if "all_models" not in globals():
    all_models = [
        ("K-Means",       labels_kmeans),
        ("Hierarchical",  labels_hc),
        ("DBSCAN",        labels_db),
        ("Spectral",      labels_spec),
        ("K-Means_8",     labels_kmeans_8),
        ("DBSCAN_7",      labels_db_7),
    ]

fig, axes = plt.subplots(3, 2, figsize=(14, 15), dpi=300)

for ax, (name, labels) in zip(axes.flat, all_models):
    sc = ax.scatter(
        X_2d[:, 0], X_2d[:, 1],
        c=labels,
        cmap="tab10",
        alpha=0.65,
        s=18,
        linewidths=0
    )
    ax.set_title(name)
    ax.set_xlabel("PC1")
    ax.set_ylabel("PC2")
    fig.colorbar(sc, ax=ax, label="Cluster")

fig.suptitle("So sánh 6 kết quả phân cụm – Không gian 2D PCA", y=1.01)
plt.tight_layout()
save_report_fig(fig, "pca_2d_cluster_comparison")
plt.show()


# ── Xuất hình so sánh cụm 3D PCA chất lượng cao ─────────────────────────
if "X_3d" not in globals():
    pca3d = PCA(n_components=3, random_state=42)
    X_3d = pca3d.fit_transform(X_model)

fig = plt.figure(figsize=(16, 18), dpi=300)

for i, (name, labels) in enumerate(all_models, 1):
    ax = fig.add_subplot(3, 2, i, projection="3d")
    sc = ax.scatter(
        X_3d[:, 0], X_3d[:, 1], X_3d[:, 2],
        c=labels,
        cmap="tab10",
        alpha=0.6,
        s=12
    )
    ax.set_title(name)
    ax.set_xlabel("PC1")
    ax.set_ylabel("PC2")
    ax.set_zlabel("PC3")
    fig.colorbar(sc, ax=ax, shrink=0.5, label="Cluster")

fig.suptitle("So sánh 6 kết quả phân cụm – Không gian 3D PCA", y=1.01)
plt.tight_layout()
save_report_fig(fig, "pca_3d_cluster_comparison")
plt.show()
```

### 10d. Xuất bảng tổng kết chỉ số thành heatmap

Cell này trực quan hóa `df_summary` để đưa vào báo cáo. Đây chỉ là hình minh họa cho các chỉ số đánh giá, không thay đổi kết quả model.


```python
# ── Heatmap tổng kết chỉ số đánh giá model ──────────────────────────────
# Dùng lại df_summary đã tính ở phần tổng kết.

if "df_summary" not in globals():
    raise ValueError("Chưa có df_summary. Hãy chạy lại phần Tổng kết – Silhouette · DBI · CHI trước.")

metrics_for_heatmap = df_summary.set_index("Algorithm")[["Silhouette", "DBI", "CHI"]].copy()

# Chuẩn hóa từng cột về [0, 1] để đưa lên cùng một heatmap dễ nhìn hơn.
# Lưu ý: đây chỉ là chuẩn hóa để trực quan hóa, không dùng để đánh giá lại model.
metrics_norm = metrics_for_heatmap.copy()
for col in metrics_norm.columns:
    col_min = metrics_norm[col].min()
    col_max = metrics_norm[col].max()
    if pd.isna(col_min) or pd.isna(col_max) or col_max == col_min:
        metrics_norm[col] = np.nan
    else:
        metrics_norm[col] = (metrics_norm[col] - col_min) / (col_max - col_min)

fig, ax = plt.subplots(figsize=(8, 5), dpi=300)
sns.heatmap(
    metrics_norm,
    annot=metrics_for_heatmap,
    fmt=".3g",
    cmap="YlGnBu",
    linewidths=0.5,
    linecolor="white",
    ax=ax
)
ax.set_title("Tổng kết chỉ số đánh giá phân cụm")
ax.set_xlabel("Chỉ số đánh giá")
ax.set_ylabel("Thuật toán")
plt.tight_layout()
save_report_fig(fig, "heatmap_model_evaluation_summary")
plt.show()
```
